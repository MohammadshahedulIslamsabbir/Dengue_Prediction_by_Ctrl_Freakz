```python
# Create a chronologically sorted copy

df_ml = df.sort_values(
    ["ward_id", "date"]
).copy()

# Reset row indexes
df_ml = df_ml.reset_index(drop=True)

# Verify sorting

print("Total Records:", len(df_ml))

print("\nFirst 10 Records:")
display(df_ml[["date", "city", "ward_id", "rainfall_mm", "dengue_cases"]].head(10))

# Check chronological order within each ward
date_check = df_ml.groupby("ward_id")["date"].apply(
    lambda x: x.is_monotonic_increasing
)

print("\nAll Wards Chronologically Sorted:", date_check.all())
```

```python
# Create rainfall lag features for each ward

for lag in range(1, 5):
    df_ml[f"rainfall_lag_{lag}"] = (
        df_ml.groupby("ward_id")["rainfall_mm"].shift(lag)
    )

# Display the results

print("Rainfall Lag Features Created!")

display(
    df_ml[
        [
            "date",
            "ward_id",
            "rainfall_mm",
            "rainfall_lag_1",
            "rainfall_lag_2",
            "rainfall_lag_3",
            "rainfall_lag_4"
        ]
    ].head(10)
)
```

```python
# Create accumulated rainfall features

df_ml["rainfall_sum_2w"] = (
    df_ml.groupby("ward_id")["rainfall_mm"]
    .transform(lambda x: x.shift(1).rolling(2).sum())
)

df_ml["rainfall_sum_3w"] = (
    df_ml.groupby("ward_id")["rainfall_mm"]
    .transform(lambda x: x.shift(1).rolling(3).sum())
)

df_ml["rainfall_sum_4w"] = (
    df_ml.groupby("ward_id")["rainfall_mm"]
    .transform(lambda x: x.shift(1).rolling(4).sum())
)

print("Multi-week Rainfall Features Created!")

display(
    df_ml[
        [
            "date",
            "ward_id",
            "rainfall_mm",
            "rainfall_sum_2w",
            "rainfall_sum_3w",
            "rainfall_sum_4w"
        ]
    ].head(10)
)
```

```python
# Create future dengue prediction targets

for weeks in [2, 3, 4]:
    df_ml[f"dengue_cases_t_plus_{weeks}"] = (
        df_ml.groupby("ward_id")["dengue_cases"].shift(-weeks)
    )

print("Future Dengue Targets Created!")

display(
    df_ml[
        [
            "date",
            "ward_id",
            "dengue_cases",
            "dengue_cases_t_plus_2",
            "dengue_cases_t_plus_3",
            "dengue_cases_t_plus_4"
        ]
    ].head(10)
)
```

```python
import numpy as np

# Create seasonal features

df_ml["month"] = df_ml["date"].dt.month

df_ml["month_sin"] = np.sin(
    2 * np.pi * df_ml["month"] / 12
)

df_ml["month_cos"] = np.cos(
    2 * np.pi * df_ml["month"] / 12
)

print("Seasonal Features Created!")

display(
    df_ml[
        ["date", "month", "month_sin", "month_cos"]
    ].head(15)
)
```

```python
# Define split boundaries

validation_start = pd.Timestamp("2024-01-01")
test_start = pd.Timestamp("2025-01-06")

# Four-week gap before each boundary

training_cutoff = validation_start - pd.Timedelta(weeks=4)
validation_cutoff = test_start - pd.Timedelta(weeks=4)

# Create datasets

train_df = df_clean[
    df_clean["date"] < training_cutoff
].copy()

validation_df = df_clean[
    (df_clean["date"] >= validation_start) &
    (df_clean["date"] < validation_cutoff)
].copy()

test_df = df_clean[
    df_clean["date"] >= test_start
].copy()

# Display dataset information

datasets = {
    "Training": train_df,
    "Validation": validation_df,
    "Testing": test_df
}

for name, data in datasets.items():
    print(f"\n{name} Dataset")
    print("-------------------------")
    print("Records:", len(data))
    print("First Date:", data["date"].min())
    print("Last Date:", data["date"].max())
    print("Unique Wards:", data["ward_id"].nunique())

# Verify no missing values

print("\nTotal Records:",
      len(train_df) + len(validation_df) + len(test_df))
```

```python
# Numeric input features

numeric_features = [
    "population_density_sqkm",
    "water_logging_risk_index",
    "sanitation_score",

    "rainfall_lag_1",
    "rainfall_lag_2",
    "rainfall_lag_3",
    "rainfall_lag_4",
    "rainfall_sum_2w",
    "rainfall_sum_3w",
    "rainfall_sum_4w",

    "avg_temp_lag_1",
    "avg_temp_lag_2",
    "avg_temp_lag_3",
    "avg_temp_lag_4",

    "max_temp_lag_1",
    "max_temp_lag_2",
    "max_temp_lag_3",
    "max_temp_lag_4",

    "humidity_lag_1",
    "humidity_lag_2",
    "humidity_lag_3",
    "humidity_lag_4",

    "dengue_lag_1",
    "dengue_lag_2",
    "dengue_lag_3",
    "dengue_lag_4",

    "month_sin",
    "month_cos"
]

# Categorical input features

categorical_features = [
    "city",
    "ward_id"
]

# Combine all input features

feature_columns = numeric_features + categorical_features

# Prediction targets

target_columns = [
    "dengue_cases_t_plus_2",
    "dengue_cases_t_plus_3",
    "dengue_cases_t_plus_4"
]

# Verify

print("Numeric Features:", len(numeric_features))
print("Categorical Features:", len(categorical_features))
print("Total Input Features:", len(feature_columns))
print("Prediction Targets:", len(target_columns))

print("\nAll Input Features:")
print(feature_columns)

print("\nAll Prediction Targets:")
print(target_columns)

# Check whether all columns exist

print("\nAll Features Exist:",
      all(col in df_clean.columns for col in feature_columns))

print("All Targets Exist:",
      all(col in df_clean.columns for col in target_columns))
```

```python
# Prepare training inputs and targets

X_train = train_df[feature_columns]
y_train = train_df[target_columns]

# Prepare validation inputs and targets

X_validation = validation_df[feature_columns]
y_validation = validation_df[target_columns]

# Prepare testing inputs and targets

X_test = test_df[feature_columns]
y_test = test_df[target_columns]

# Display shapes

print("Training Data:")
print("X_train:", X_train.shape)
print("y_train:", y_train.shape)

print("\nValidation Data:")
print("X_validation:", X_validation.shape)
print("y_validation:", y_validation.shape)

print("\nTesting Data:")
print("X_test:", X_test.shape)
print("y_test:", y_test.shape)
```

```python
##pipeline
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder
from sklearn.pipeline import Pipeline
from sklearn.ensemble import RandomForestRegressor

# 1. Create preprocessing component

preprocessor = ColumnTransformer(
    transformers=[
        ("numeric", "passthrough", numeric_features),
        ("categorical",
         OneHotEncoder(handle_unknown="ignore"),
         categorical_features)
    ]
)

# 2. Create Random Forest model

rf_model = RandomForestRegressor(
    n_estimators=200,
    random_state=42,
    n_jobs=-1,
    min_samples_leaf=2
)

# 3. Create three separate prediction pipelines

model_pipelines = {}

for target in target_columns:

    model_pipelines[target] = Pipeline(
        steps=[
            ("preprocessor", preprocessor),
            ("model", rf_model)
        ]
    )

# 4. Verify pipeline creation

print("Preprocessing Pipeline:", type(preprocessor).__name__)
print("ML Algorithm:", type(rf_model).__name__)
print("Number of Models:", len(model_pipelines))

print("\nCreated Models:")

for model_name in model_pipelines:
    print(model_name)

print("\nPipeline Creation Successful!")
```

```python
from sklearn.base import clone

# Create three independent pipelines

model_pipelines = {}

for target in target_columns:

    # Create a fresh Random Forest for each target
    rf_model = RandomForestRegressor(
        n_estimators=200,
        random_state=42,
        n_jobs=-1,
        min_samples_leaf=2
    )

    # Create an independent preprocessing pipeline
    model_pipelines[target] = Pipeline(
        steps=[
            ("preprocessor", clone(preprocessor)),
            ("model", rf_model)
        ]
    )

# Verify that all three models are independent

print("Number of Models:", len(model_pipelines))

print(
    "Are all model objects independent?",
    len({id(pipeline.named_steps["model"])
         for pipeline in model_pipelines.values()}) == 3
)

print(
    "Are all preprocessors independent?",
    len({id(pipeline.named_steps["preprocessor"])
         for pipeline in model_pipelines.values()}) == 3
)

print("\nAll three independent pipelines are ready!")
```

```python
##Model traning
```

```python
import time

# Store training time for each model
training_times = {}

print("Starting Random Forest Model Training...\n")

for target in target_columns:

    print(f"Training {target}...")

    start_time = time.time()

    # Train the model
    model_pipelines[target].fit(
        X_train,
        train_df[target]
    )

    end_time = time.time()

    # Calculate training time
    elapsed_time = end_time - start_time

    training_times[target] = elapsed_time

    print(f"Completed in {elapsed_time:.2f} seconds\n")


print("All Three Models Trained Successfully!")

print("\nTraining Time Summary:")

for target, elapsed_time in training_times.items():
    print(f"{target}: {elapsed_time:.2f} seconds")
```

```python
# Generate predictions for validation data

validation_predictions = {}

print("Generating Validation Predictions...\n")

for target in target_columns:

    predictions = model_pipelines[target].predict(
        X_validation
    )

    validation_predictions[target] = predictions

    print(f"{target}:")
    print("Total Predictions:", len(predictions))
    print("First 10 Predictions:", predictions[:10])
    print()

print("All Validation Predictions Generated!")
```

```python
from sklearn.metrics import mean_absolute_error
from sklearn.metrics import mean_squared_error
from sklearn.metrics import r2_score
import numpy as np

evaluation_results = []

print("Random Forest Validation Evaluation\n")

for target in target_columns:

    # Actual values
    actual = y_validation[target]

    # Predicted values
    predicted = validation_predictions[target]

    # Random Forest metrics
    mae = mean_absolute_error(actual, predicted)

    rmse = np.sqrt(mean_squared_error(actual, predicted))

    r2 = r2_score(actual, predicted)

    # Baseline: always predict training mean
    baseline_prediction = np.full(
        len(actual),
        y_train[target].mean()
    )

    baseline_mae = mean_absolute_error(
        actual,
        baseline_prediction
    )

    baseline_rmse = np.sqrt(
        mean_squared_error(actual, baseline_prediction)
    )

    evaluation_results.append({
        "Prediction Horizon": target,
        "RF MAE": mae,
        "RF RMSE": rmse,
        "RF R2": r2,
        "Baseline MAE": baseline_mae,
        "Baseline RMSE": baseline_rmse
    })

# Display results
evaluation_df = pd.DataFrame(evaluation_results)

display(evaluation_df.round(4))
```

```python
# Compare Random Forest with previous-week baseline

persistence_results = []

for target in target_columns:

    actual = y_validation[target]

    # Use the most recent observed dengue count
    previous_week_prediction = X_validation["dengue_lag_1"]

    # Calculate baseline metrics
    persistence_mae = mean_absolute_error(
        actual,
        previous_week_prediction
    )

    persistence_rmse = np.sqrt(
        mean_squared_error(actual, previous_week_prediction)
    )

    # Random Forest metrics
    rf_prediction = validation_predictions[target]

    rf_mae = mean_absolute_error(actual, rf_prediction)

    persistence_results.append({
        "Prediction Horizon": target,
        "RF MAE": rf_mae,
        "Persistence MAE": persistence_mae,
        "RF RMSE": np.sqrt(mean_squared_error(actual, rf_prediction)),
        "Persistence RMSE": persistence_rmse
    })

persistence_df = pd.DataFrame(persistence_results)

display(persistence_df.round(4))
```
